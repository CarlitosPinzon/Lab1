<h2 id="creating">Crear una Clase</h2>

> **Nota propia:** La clase se declara con class seguido del nombre. El método __init__ es el constructor: se ejecuta automáticamente al crear el objeto y sirve para inicializar los atributos. El parámetro self hace referencia al objeto que se está creando, y por eso aparece primero en todos los métodos.

Creemos ahora una clase circle, pero primero, importemos una librería para dibujar objetos:

```python
# Importar la librería

import matplotlib.pyplot as plt
%matplotlib inline
```

El primer paso en la creación de nuestra clase es usar la palabra reservada <code>class</code>, y después el nombre de la clase como se muestra en la figura 4. En este curso la clase padre será siempre objeto:

El siguiente paso es el método especial llamado constructor <code>&#95;&#95;init&#95;&#95;</code>, que es usado para inicializar el objeto. La entrada son datos para los atributos. El termino <code>self</code> contiene todos los atributos del grupo. Por ejemplo <code>self.color</code> asigna el valor del atributo color y <code>self.radius</code> valor del radio. También tenemos el método <code>add_radius()</code> con el parámetro <code>r</code>, el método suma el valor de <code>r</code> al atributo radius. Para acceder al valor de radius usamos la sintaxis <code>self.radius</code>. El etiquetado se resume en la Figura 5.

<img src="https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%203/Images/ClassesCircle.png" width="600" />

El objeto actual se muestra mas adelante. Incluimos el método <code>drawCircle</code> para mostrar la imagen de un circulo. Asignamos el valor de radius en 3 y el color en azul (blue):

```python
# Crear la clase Circle

class Circle(object):

    # Constructor
    def __init__(self, radius=3, color='blue'):
        self.radius = radius
        self.color = color

    # Method
    def add_radius(self, r):
        self.radius = self.radius + r
        return(self.radius)

    # Method
    def drawCircle(self):
        plt.gca().add_patch(plt.Circle((0, 0), radius=self.radius, fc=self.color))
        plt.axis('scaled')
        plt.show()
```

<h2 id="rect">La Clase Rectángulo</h2>

> **Nota propia:** La clase Rectangulo repite la misma estructura con otros atributos, lo que confirma que el patrón de constructor más métodos es general. Anoto que en el cuaderno original el objeto FatYellowRectangle se creaba como Rectangle(...) cuando la clase se llama Rectangulo; lo corregí para que la celda pueda ejecutarse.

Vamos a crear una clase rectángulo con los atributos height (altura) y color. Solo añadiremos el método para dibujar un rectángulo:

```python
# Crear una nueva clase Rectangle para construir un objeto rectangulo

class Rectangulo(object):

    # Constructor
    def __init__(self, width=2, height=3, color='r'):
        self.height = height
        self.width = width
        self.color = color

    # Method
    def drawRectangle(self):
        plt.gca().add_patch(plt.Rectangle((0, 0), self.width, self.height ,fc=self.color))
        plt.axis('scaled')
        plt.show()
```

Creamos el objeto <code>SkinnyBlueRectangle</code> de tipo Rectangle. Su anchura (width) será de 2 y la altura (height) de 3, será de color azul:

```python
# Crear un nuevo objeto rectangle

SkinnyBlueRectangle = Rectangulo(2, 10, 'blue')
```

La clase matplotlib.patches.Ellipse, permite dibujar elipses desde [matplotlib](https://matplotlib.org/stable/api/_as_gen/matplotlib.patches.Ellipse.html). La clase se importa de la siguiente manera:

*from matplotlib.patchs import Ellipse*

Se puede dibujar entonces cualquier elipse, dando los argumentos:

plt.gca().add_patch(Ellipse(xy, width, height, facecolor = color, edgecolor = edgecolor))

Se requiere crear una clase Elipse, la cual el constructor sea con valores iniciales cualesquiera. Adicionalmente, la clase debe tener los siguientes métodos:

1.- Cambiar el ancho

2.- Cambiar el alto

3.- Cambiar el color de relleno

4.- Cambiar el color de borde

5.- Dibujar

> **Nota propia:** Para la actividad sigo la misma estructura de las clases anteriores. En el constructor dejo valores iniciales por defecto para poder crear una elipse sin pasar argumentos. Los cuatro métodos de cambio simplemente reasignan el atributo correspondiente, y el método dibujar usa Ellipse de matplotlib pasándole el centro, el ancho, el alto y los dos colores. Uso return self en los métodos de cambio para poder encadenar llamadas.

```python
# Solución de la actividad: clase Elipse

from matplotlib.patches import Ellipse


class Elipse(object):

    # Constructor con valores iniciales
    def __init__(self, xy=(0, 0), width=4, height=2, color='cyan', edgecolor='black'):
        self.xy = xy
        self.width = width
        self.height = height
        self.color = color
        self.edgecolor = edgecolor

    # 1. Cambiar el ancho
    def cambiar_ancho(self, width):
        self.width = width
        return self

    # 2. Cambiar el alto
    def cambiar_alto(self, height):
        self.height = height
        return self

    # 3. Cambiar el color de relleno
    def cambiar_color(self, color):
        self.color = color
        return self

    # 4. Cambiar el color de borde
    def cambiar_borde(self, edgecolor):
        self.edgecolor = edgecolor
        return self

    # 5. Dibujar
    def dibujar(self):
        plt.gca().add_patch(Ellipse(self.xy, self.width, self.height,
                                    facecolor=self.color,
                                    edgecolor=self.edgecolor))
        plt.axis('scaled')
        plt.show()
```

Creo una primera elipse usando únicamente los valores por defecto del constructor:

```python
# Elipse con los valores por defecto

elipse1 = Elipse()
print('Ancho:', elipse1.width, '| Alto:', elipse1.height, '| Color:', elipse1.color)
elipse1.dibujar()
```

Ahora creo una segunda elipse con valores propios y compruebo cada uno de los métodos de cambio:

```python
# Elipse con valores propios

elipse2 = Elipse(xy=(0, 0), width=8, height=3, color='orange', edgecolor='red')
elipse2.dibujar()
```

```python
# Probar los métodos de cambio uno por uno

elipse2.cambiar_ancho(5)
elipse2.cambiar_alto(5)
elipse2.cambiar_color('lightgreen')
elipse2.cambiar_borde('darkgreen')

print('Ancho:', elipse2.width)
print('Alto:', elipse2.height)
print('Color de relleno:', elipse2.color)
print('Color de borde:', elipse2.edgecolor)

elipse2.dibujar()
```

Como los métodos devuelven el propio objeto, también puedo encadenarlos en una sola línea:

```python
# Encadenar los métodos en una sola instrucción

Elipse().cambiar_ancho(10).cambiar_alto(2).cambiar_color('violet').cambiar_borde('purple').dibujar()
```
